```python
# 1. Temperature conversion

def convert_temperature(temp, unit):
    """Convert F -> C"""
    if unit == 'C':
        return (temp * (9/5))+32
    #Convert C -> F
    elif unit == "F":
        return ((temp - 32)*(5/9))
    else:
        return None

print(convert_temperature(100,"F"))
print(convert_temperature(100,"C"))

print(convert_temperature(212,"F"))
print(convert_temperature(37.77777777777778,"C"))


```

```python
# Password strenght checker

def is_strong_password(password):
    if len(password)<8:
        return False
    if not any(char.isdigit() for char in password):# this checks if it is number
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&*()_+' for char in password):
        return False
    return True
    #any() checks a collection of values and returns True if at least one value is truthy.



    
```

```python
print(is_strong_password("WeakPaswesd"))
```

```python
print(is_strong_password("He110@World"))
```

```python
cart=[
    {'name':'Apple','price':0.5,'quantity':4},
    {'name':'Banana','price':0.3,'quantity':6},
    {'name':'Orange','price':0.7,'quantity':3}

]

def total_cost_of_items(cart):
    totalcost= 0
    for item in cart:
        totalcost += (item["price"] * item["quantity"])
    return totalcost

print(total_cost_of_items(cart))
```

```python
# check if the value is palindrome
def palindrome_checker(text):
    text == text.lower().replace(" ","")
    return text == text[::-1]
    
print(palindrome_checker("aBa"))
     
```
