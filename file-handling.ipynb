```python
with open('source.txt', mode='r') as file:
    content = file.read()
    print(content)
```

```python
# Read file line by line

with open ('source.txt', mode='r') as file:
    for line in file:
        print(line)
        print(line.strip())
```

```python
# writing into file - overwriting

with open('source.txt', 'w') as file:
    file.write('Hello World!\n')
    file.write("How are you?")
```

```python
# writing into file without overwriting

with open('source.txt', mode='a') as file:
    file.write("\nI am learning and practicing Python concepts!")
```

```python
## Writing list of lines "file.writelines"

lines_list = ["\nFirst Line", "Second Line\n", "Third Line"]

with open('source.txt', mode = 'a') as file:
    file.writelines(lines_list)
```
