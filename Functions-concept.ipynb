```python
#Variable lenght argument - positional argument
# these are postional arguments that can be used if the number of arguments are not fixed.

def print_numbers(*abc):
    for numbers in abc:
        print(numbers)

print_numbers(1,2,3,"Hello", True, 1.5)
```

```python
## keyword arguments
# mostly can use key value pairs
def print_keyword_ar_function(**abc):
    for key,value in abc.items():
        print(f"{key}:{value}")

print_keyword_ar_function(name="Sachin", age="32", country="India")
```

```python
print_keyword_ar_function(1,2,3,name="Sachin", age="32", country="India")
```

```python
# both positional and keyword argument

def post_keyword_arg_func(*abc, **xyz):
    for val in abc:
        print(f"Positonal argument values: {val}")

    for key,value in xyz.items():
        print(f"{key}: {value}")



post_keyword_arg_func(1,2,3,"Hello",True, name="Sachin", age=23, country="India")
```

```python
# multiple return values

def multi_return(a,b):
    return a*b, a, b

multi_return(10,20)
```
