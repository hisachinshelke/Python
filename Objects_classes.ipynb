```python
class Library:
    def __init__(self, bookName, issueStatus=False, issuedBy=None, issueDate=None):
        self.bookName = bookName
        self.issueStatus = issueStatus
        self.issuedBy = issuedBy
        self.issueDate = issueDate

    def getBookStatus(self, bookName):
        if bookName in self.bookName:
            return "Book is available"
        else:
            print(f"Book is not availalble. It is issused by {self.issuedBy}.")

    def registerBook(self, bookName):
        self.bookName.append(bookName)

    def issueBook(self, bookName, issuedBy, issueDate):
        self.bookName.remove(bookName)
        self.issuedBy =issuedBy
        self.issueDate = issueDate
        self.issueStatus = True

    
    def returnBook(self, bookName, issuedBy, issueDate):
        self.bookName.append(bookName)
        self.issuedBy =None
        self.issueDate = None
        self.issueStatus = False

```

```python
bookAbc = Library("ABC")
```

```python
bookAbc.getBookStatus("Abc")
```
