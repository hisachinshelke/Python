```python
## CSV library

import csv

with open("examples.csv", mode='w', newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["name","age"])
    writer.writerow(["Sachin","33"])
    writer.writerow(["Kranti","28"])


```

```python
with open('examples.csv', mode='r') as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

```python
# date time

from datetime import datetime, timedelta

now = datetime.now()
print(now)

yesterday = now - timedelta(days=1)
print(yesterday)
```

```python
# sleep concept

import time

print(time.time())

time.sleep(5)

print(time.time())
```

```python
# Regular expression

import re

pattern = r'\d+'

text = "There are 12 apples in a dozen. 6 in half dozen"

match = re.search(pattern, text)

print(match)
print(match.group())
```
